```python
def Cov_Analytical(t, s, n_p, Gamma_x):
    term1 = (1+2*n_p)*np.array([[s*np.cos(t)**2 + np.sin(t)**2/s,(1/s-s)*np.cos(t)*np.sin(t)],
                                [(1/s-s)*np.cos(t)*np.sin(t),s*np.sin(t)**2 + np.cos(t)**2/s]])
    
    term2 = Gamma_x*np.array([[t - np.cos(t)*np.sin(t),np.sin(t)**2],
                               [np.sin(t)**2,t + np.cos(t)*np.sin(t)]])
    return term1+term2

Cov_array = Cov_Analytical(t_array, s, N0, gamma_x)

plt.plot(t_array, GCS.sigma_JK_t[:,1,0,1,1])
plt.plot(t_array, Cov_array[1,1], ':')
plt.plot(t_array, GCS.sigma_JK_t[:,1,0,0,0])
plt.plot(t_array, Cov_array[0,0], ':')
plt.plot(t_array, GCS.sigma_JK_t[:,1,0,1,0])
plt.plot(t_array, Cov_array[0,1], ':')
```

```python
def r_minus(tau, f_q, f_u):

    cos_tau = np.cos(tau)
    sin_tau = np.sin(tau)

    # First component
    first_component = -(cos_tau - 1) 

    # Second component
    second_component = sin_tau 

    # Assemble vector
    vector =  (- f_q + f_u)* np.array([first_component, second_component])
    return vector


def r_plus(tau, f_q, f_u):

    cos_tau = np.cos(tau)
    sin_tau = np.sin(tau)

    # First component
    first_component = -(cos_tau - 1) 

    # Second component
    second_component = sin_tau 

    # Assemble vector
    vector =  (f_q + f_u) * np.array([first_component, second_component])
    return vector


r_minus_array = r_minus(t_array, f_q, f_u)
r_plus_array = r_plus(t_array, f_q, f_u)


plt.plot(t_array, GCS.r_JK_t[:,1,1,0])
plt.plot(t_array, r_plus_array[0,:], ':',color='m')
plt.plot(t_array, GCS.r_JK_t[:,1,1,1])
plt.plot(t_array, r_plus_array[1,:], ':')
```

```python
def r_off(t, s,  fq, f_u, n_p, gamma_x):
    
    real = f_u*np.array([[1 - np.cos(t)],
                        [np.sin(t)]])
    
    complex1 = (1+2*n_p)*np.array([[np.sin(t)*((1/s-s)*np.cos(t) - 1/s)],
                                 [2*np.sin(t/2)**2*(s + (s-1/s)*np.cos(t))]])
    
    complex2 = gamma_x*np.array([[-4*np.sin(t/2)**4],
                                [t + np.sin(t)*(np.cos(t) - 2)]])

    return real+1j*fq*(complex1+complex2)

r_off_array = r_off(t_array, s,  f_q, f_u, N0, gamma_x)

plt.plot(t_array, np.imag(GCS.r_JK_t[:,0,1,0]))
plt.plot(t_array, np.imag(r_off_array[0,0,:]), ':',color='m')
plt.plot(t_array, np.imag(GCS.r_JK_t[:,0,1,1]))
plt.plot(t_array, np.imag(r_off_array[1,0,:]), ':')
```

```python
def C_anlytical(t, fq, s, np_val, gamma_x, gamma_z):    

    term1 = 2*(2 * np_val + 1)*np.sin(t/2)**2*((s - 1/s)*np.cos(t) +s + 1/s)
    term2 = gamma_x*(3*t+ np.sin(t)*(np.cos(t)-4))
    term3 = gamma_z*t/2

    return fq**2*(term1 + term2) + term3

C_array = C_anlytical(t_array, f_q, s, N0, gamma_x, gamma_s)
plt.plot(t_array, GCS.C_JK_t[:,1,0])
plt.plot(t_array, C_array, ':')
```

```python
sqrtc = np.emath.sqrt          # complex-safe square root
 
 
def omega_pm(kappa, mu, lam):
    """Branch frequencies (omega_+, omega_-)."""
    op = ((kappa + 2 * lam) ** 2 + 4j * kappa * mu - 4 * mu ** 2) / 4
    om = ((kappa - 2 * lam) ** 2 + 4j * kappa * mu - 4 * mu ** 2) / 4
    return op, om
 
 
def Cdot_rbar(tau, kappa, mu, lam, alpha, phi_d):
    """
    Drive-induced decoherence rate: minus the real part of
 
      -8 i alpha^2 kappa mu [ cos^2(phi_d) ( (kappa+2lam+2i mu)/(2 sqrt(w+))
                                             [cosh(tau sqrt(w+)) - 1]
                                             + sinh(tau sqrt(w+)) )^2
                              / ( (kappa+2lam+2i mu) sinh(tau sqrt(w+))
                                  + 2 sqrt(w+) cosh(tau sqrt(w+)) )^2
                            + sin^2(phi_d) ( ... same with w-, kappa-2lam ... ) ]
    """
    tau = np.asarray(tau, dtype=float)
    op, om = omega_pm(kappa, mu, lam)
    total = 0.0
    for w, A, trig in ((op, kappa + 2 * lam + 2j * mu, np.cos(phi_d) ** 2),
                       (om, kappa - 2 * lam + 2j * mu, np.sin(phi_d) ** 2)):
        r = sqrtc(w)
        num = A / (2 * r) * (np.cosh(tau * r) - 1) + np.sinh(tau * r)
        den = A * np.sinh(tau * r) + 2 * r * np.cosh(tau * r)
        total = total + trig * (num / den) ** 2
    return -np.real(-8j * alpha ** 2 * kappa * mu * total)
 
 
def Cdot_sigma(tau, kappa, mu, lam):
    """
    Parasitic decoherence rate: minus the real part of
 
      -(i mu / 2) [ ( (kappa-2lam+2i mu) sinh(tau sqrt(w+))
                      + 2 sqrt(w+) cosh(tau sqrt(w+)) )
                    / ( (kappa+2lam+2i mu) sinh(tau sqrt(w+))
                        + 2 sqrt(w+) cosh(tau sqrt(w+)) )
                  + ( (kappa+2lam+2i mu) sinh(tau sqrt(w-))
                      + 2 sqrt(w-) cosh(tau sqrt(w-)) )
                    / ( (kappa-2lam+2i mu) sinh(tau sqrt(w-))
                        + 2 sqrt(w-) cosh(tau sqrt(w-)) ) ]
    """
    tau = np.asarray(tau, dtype=float)
    op, om = omega_pm(kappa, mu, lam)
    ap, am = kappa + 2 * lam + 2j * mu, kappa - 2 * lam + 2j * mu
    out = 0.0
    for w, num_c, den_c in ((op, am, ap), (om, ap, am)):
        r = sqrtc(w)
        sh, ch = np.sinh(tau * r), np.cosh(tau * r)
        out = out + (num_c * sh + 2 * r * ch) / (den_c * sh + 2 * r * ch)
    return -np.real(-0.5j * mu * out)
 
 
# ----------------------------------------------------------------------
# tau -> infinity limits, in closed form
# ----------------------------------------------------------------------
def Cdot_sigma_inf(kappa, mu, lam):
    """Re[ sqrt(w+) + sqrt(w-) ] / 2 - kappa/2  ==  Eq. (2) without 1/T2*."""
    op, om = omega_pm(kappa, mu, lam)
    return np.real(sqrtc(op) + sqrtc(om)) / 2 - kappa / 2
 
 
def Cdot_rbar_inf(kappa, mu, lam, alpha, phi_d):
    """2 mu^2 kappa^2 alpha^2 [ cos^2(phi_d)/|w+|^2 + sin^2(phi_d)/|w-|^2 ]
    == the drive term of Eq. (4)."""
    op, om = omega_pm(kappa, mu, lam)
    return 2 * mu ** 2 * kappa ** 2 * alpha ** 2 * (
        np.cos(phi_d) ** 2 / abs(op) ** 2 + np.sin(phi_d) ** 2 / abs(om) ** 2)
 
 
def eddins_eq2(kappa, mu, lam):
    """Eq. (2) as printed: (1/2) Re[ sqrt(D(-lam)) + sqrt(D(lam)) ] - kappa/2,
    with D(l) = (kappa/2 + l + i mu)^2 - 2 i mu l."""
    D = lambda l: (kappa / 2 + l + 1j * mu) ** 2 - 2j * mu * l
    return 0.5 * np.real(sqrtc(D(-lam)) + sqrtc(D(lam))) - kappa / 2
 
 
def lambda_from_gain(kappa, gain_dB):
    """lambda from the QPA power gain: sqrt(G) = (kappa/2 + lam)/(kappa/2 - lam)."""
    G = 10 ** (gain_dB / 10)
    return (kappa / 2) * (np.sqrt(G) - 1) / (np.sqrt(G) + 1)




kappa, mu, alpha = 25.4, 1.9, 0.9
print(f"kappa/2pi = {kappa} MHz,  mu/2pi = {mu} MHz,  alpha = {alpha}\n")

print("=== 1. omega_+- coincide with D(+-lambda) of the reference ===")
lam = lambda_from_gain(kappa, 3.0)
op, om = omega_pm(kappa, mu, lam)
D = lambda l: (kappa / 2 + l + 1j * mu) ** 2 - 2j * mu * l
print(f"  G = 3 dB -> lambda = {lam:.4f}")
print(f"  omega_+ = {op:.6f}   D(+lambda) = {D(lam):.6f}   |diff| = {abs(op - D(lam)):.1e}")
print(f"  omega_- = {om:.6f}   D(-lambda) = {D(-lam):.6f}   |diff| = {abs(om - D(-lam)):.1e}")

print("\n=== 2. Cdot_sigma: transient -> its closed-form limit == Eq. (2) ===")
print("  G[dB]  lambda    kt=5        kt=20       kt=60      limit        Eq.(2)      |diff|")
for g in [0.0, 1.0, 3.0, 6.0, 10.0]:
    lam = lambda_from_gain(kappa, g)
    v = [Cdot_sigma(x / kappa, kappa, mu, lam) for x in (5, 20, 60)]
    lim, e2 = Cdot_sigma_inf(kappa, mu, lam), eddins_eq2(kappa, mu, lam)
    print(f"  {g:5.1f} {lam:7.4f}  {v[0]:.8f}  {v[1]:.8f}  {v[2]:.8f}  "
          f"{lim:.8f}  {e2:.8f}  {abs(lim - e2):.1e}")

print("\n=== 3. Cdot_rbar: transient -> its closed-form limit == drive term of Eq. (4) ===")
lam = lambda_from_gain(kappa, 3.0)
print("  mode        kt=5        kt=20       kt=60      limit (Eq. 4)   |kt=60 - limit|")
for phi_d, name in [(0.0, "amplified"), (np.pi / 2, "squeezed ")]:
    v = [Cdot_rbar(x / kappa, kappa, mu, lam, alpha, phi_d) for x in (5, 20, 60)]
    lim = Cdot_rbar_inf(kappa, mu, lam, alpha, phi_d)
    print(f"  {name}  {v[0]:.8f}  {v[1]:.8f}  {v[2]:.8f}   {lim:.8f}    {abs(v[2] - lim):.1e}")

print("\n=== 4. full transient, G = 3 dB, phi_d = 0 ===")
print("  kappa*tau    Cdot_rbar     Cdot_sigma       total")
for x in [0.5, 1, 2.5, 5, 10, 25, 50]:
    t = x / kappa
    a = Cdot_rbar(t, kappa, mu, lam, alpha, 0.0)
    b = Cdot_sigma(t, kappa, mu, lam)
    print(f"  {x:8.1f}   {a:.8f}    {b:.8f}    {a + b:.8f}")
 
```

```python
#!/usr/bin/env python3
"""Rates figure in the style of plot_contrasts()."""
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib import rcParams


kappa, mu, alpha = 25, 2, 1      # 2*pi*MHz, 2*pi*MHz, input drive amplitude
gain_dB = 3.0


def plot_rates(kt_array, save):
    """
    Decoherence rates against time.
      blue  -> no squeezing (lambda = 0)
      red   -> squeezed quadrature   (phi_d = pi/2, gain on)
      green -> amplified quadrature  (phi_d = 0,    gain on)
      solid  -> Cdot_rbar  (first moments, drive-induced)
      dashed -> Cdot_sigma (covariance, parasitic)
    kt_array is kappa*tau (dimensionless).
    """
    fig, ax = plt.subplots(1, 1, figsize=(3, 2), dpi=300)
    inkscape_colors = ['#93d336ff', '#8b8ae0ff', '#f41a048a']   # green, blue, red
    rcParams['mathtext.fontset'] = 'cm'
    rcParams['font.family'] = 'STIXGeneral'

    tau = kt_array / kappa
    lam_on = lambda_from_gain(kappa, gain_dB)

    cases = [(0.0,     0.0,        inkscape_colors[1]),   # no squeezing
             (lam_on,  np.pi / 2,  inkscape_colors[2]),   # squeezed
             (lam_on,  0.0,        inkscape_colors[0])]   # amplified

    for lam, phi_d, col in cases:
        ax.plot(kt_array, Cdot_rbar(tau, kappa, mu, lam, alpha, phi_d),
                color=col, linestyle='-')
    # Cdot_sigma does not depend on phi_d, so the squeezed and amplified cases
    # share a single dashed curve; lambda = 0 gives identically zero.
    for i, (lam, col, ls) in enumerate([(lam_on, inkscape_colors[2], '-'),
                                        (lam_on, inkscape_colors[0], '--'),
                                        (0.0,    inkscape_colors[1], '--')]):
        ax.plot(kt_array, Cdot_sigma(tau, kappa, mu, lam),
                color=col, linestyle=ls, zorder=3 + i)
        
    ax.plot([-9, -8], [1, 1], '-', color=inkscape_colors[1], label=r'No Squeezing')
    ax.plot([-9, -8], [1, 1], '-', color=inkscape_colors[2], label=r'Squeezed')
    ax.plot([-9, -8], [1, 1], '-', color=inkscape_colors[0], label=r'Amplified')
    ax.plot([-9, -8], [1, 1], '-', color='gray', label=r'$\Gamma_{\bar{r}}$')
    ax.plot([-9, -8], [1, 1], '--', color='gray', label=r'$\Gamma_{\sigma}$')

    ax.grid(zorder=0, linewidth=0.6)
    ax.set_xlim(-0.5, kt_array[-1] + 0.5)
    ax.set_ylim(-0.005, 0.30)
    ax.set_ylabel(r'Rates $(\Gamma = \dot{\mathcal{C}})$', fontsize=10)
    ax.set_xlabel(r'Time ($\kappa \tau$)', fontsize=10)
    ax.set_axisbelow(True)
    ax.set_xticks([0, 5, 10, 15, 20])
    fig.suptitle('Time Evolution of Decoherence Rates', y=1.05)

    handles, labels = ax.get_legend_handles_labels()
    leg = fig.legend(handles, labels, loc='upper right', bbox_to_anchor=(1.22, 0.9),
                     fontsize=8, labelspacing=0.5, handlelength=1.2,
                     borderpad=0.2, ncols=1)
    for line in leg.get_lines():
        line.set_linewidth(1.1)
    leg.get_frame().set_linewidth(0.4)
    leg.get_frame().set_alpha(None)
    leg.get_frame().set_facecolor((1, 1, 1, 1.))

    if save is False:
        return fig
    plt.savefig(save, bbox_inches='tight')
    return fig



kt = np.linspace(1e-4, 20, 800)

lam = lambda_from_gain(kappa, gain_dB)
print("lambda(3 dB) =", round(lam, 4))
print("Cdot_sigma is phi_d-independent -> the red and green dashed curves coincide:")
for x in [5, 15, 30]:
    t = x / kappa
    print(f"   kt={x:4.0f}  sigma(phi=0)={Cdot_sigma(t,kappa,mu,lam):.10f}"
            f"   sigma(phi=pi/2)={Cdot_sigma(t,kappa,mu,lam):.10f}"
            f"   sigma(lam=0)={Cdot_sigma(t,kappa,mu,0.0):.2e}")

plot_rates(kt, "Example_1_Rates.svg")
```

## Wigner Function of the $\chi^{(1)}$ Decay Correction (Example 2)

```python
import sys, os
from gbcs import Wigner_Functions
from gbcs import Plots_Functions
from scipy.special import jv as bessel_jv


def Wigner_chi1_decay(GCS, pi_index, r_tilde_array, f_u, p_plus,
                      Gamma_1, z, phi, N_truncation, Gamma=None, branch_index=0):
    """
    Wigner function of the chi^(1) decay correction term.

    Obtained by analytically Fourier-transforming the characteristic function:

        chi^(1)(tau, r_tilde) = p_+ Gamma_1
            * exp(-1/4 r_tilde^T sigma(pi) r_tilde + i r_tilde^T r_bar_Gamma(pi))
            * S(z, phi, Gamma_1, N_truncation)

    with r_bar_Gamma(pi) = (2 f_u, 0)^T and

        S = J_0(z)(1 - exp(-Gamma pi))
          + sum_{n=1}^{N_trunc} J_n(z) * 2 i^n * [(-1)^n - exp(-Gamma_1 pi)]
            / (1 + n^2/Gamma_1^2) * [cos(n phi) + n sin(n phi) / Gamma_1]

    The Gaussian factor in r_tilde transforms analytically to the standard
    Gaussian Wigner kernel (same convention as Wigner_Sum_Diagonal).

    Parameters
    ----------
    GCS          : Quantum_State — must have sigma_JK_t and rho_q_t populated.
    pi_index     : int   — time index closest to tau = pi.
    r_tilde_array: (steps, steps, 2n, 1) — real phase-space grid.
    f_u          : float — unknown-force parameter (sets r_bar_Gamma = (2 f_u, 0)).
    p_plus       : float — population of the + branch: Re[rho_q[branch, branch]].
    Gamma_1      : float — decay rate appearing in the series denominators and exponents.
    z            : float — argument of the Bessel functions J_n(z).
    phi          : float — angle argument in cos(n phi) / sin(n phi).
    N_truncation : int   — maximum n in the series sum.
    Gamma        : float — decay rate in the J_0 term (defaults to Gamma_1).
    branch_index : int   — diagonal branch index for sigma(pi) (default 0 = + branch).

    Returns
    -------
    W1 : (steps, steps) real array
    """
    if Gamma is None:
        Gamma = Gamma_1

    n_modes = GCS.n_modes
    steps   = len(r_tilde_array)

    # Covariance matrix at tau=pi for the selected diagonal branch (real by symmetry)
    sigma_pi  = np.real(GCS.sigma_JK_t[pi_index, branch_index, branch_index])
    r_bar_Gamma = np.array([[2.0 * f_u], [0.0]])

    det_sigma = np.linalg.det(sigma_pi)
    sigma_inv = np.linalg.inv(sigma_pi)

    # Bessel series S — complex due to i^n; take Re at the end
    S = bessel_jv(0, z) * (1.0 - np.exp(-Gamma * np.pi))
    for n in range(1, N_truncation + 1):
        S += (bessel_jv(n, z)
              * 2.0 * (1j ** n)
              * ((-1) ** n - np.exp(-Gamma_1 * np.pi))
              / (1.0 + n ** 2 / Gamma_1 ** 2)
              * (np.cos(n * phi) + n * np.sin(n * phi) / Gamma_1))

    # Gaussian prefactor (same convention as Wigner_Sum_Diagonal / Wigner_Off_Diagonal)
    prefactor = (p_plus * Gamma_1 * np.real(S)
                 * (2 ** n_modes) / (np.pi ** n_modes * np.sqrt(det_sigma)))

    W1 = np.zeros((steps, steps))
    for m in range(steps):
        for nn in range(steps):
            dr = r_tilde_array[m, nn] - r_bar_Gamma
            W1[m, nn] = prefactor * float(np.exp(-(dr.T @ sigma_inv @ dr)[0, 0]))

    return W1


# ── Parameters ──────────────────────────────────────────────────────────────
# GCS here is from Example 2 (cell above); t_array = linspace(0, 2pi, 101)
_f_u      = 0.5          # unknown force (original Example 2 value)
_gamma_x  = 0.05         # Gamma_1 = position decoherence rate
_z        = 0.5          # Bessel argument — tune to physical setup
_phi      = 0.0          # Bessel phase
_N_trunc  = 20           # truncation order

# Index of tau = pi in the time array
_pi_index = np.argmin(np.abs(GCS.t_array - np.pi))

# p_+ = Re[rho_q_{00}(pi)]  (population of branch 0 at tau=pi)
_p_plus   = float(np.real(GCS.rho_q_t[_pi_index, 0, 0]))

# Phase-space grid (same sigma_para/steps as Wigner animations above)
_steps      = 101
_mode_number = 0
_sigma_para  = 0.6
_array_tick  = np.linspace(-5, 5, 3)

_r_tilde_array, _x_array, _p_array, _, _ = Wigner_Functions.Create_r_tilde_array(
    GCS, _sigma_para, _mode_number, _steps)

W1 = Wigner_chi1_decay(
    GCS, _pi_index, _r_tilde_array,
    f_u=_f_u, p_plus=_p_plus,
    Gamma_1=_gamma_x, z=_z, phi=_phi,
    N_truncation=_N_trunc)

Plots_Functions.Plot_Wigner_GCS_Total(
    W1, _x_array, _p_array, _array_tick,
    title=r'$\mathcal{W}^{(1)}(\tau{=}\pi)$ — Decay Correction',
    save=False)
```

```python
def Partial_Transpose_Dense(rho, mask, dimensions_list):
        """Computes the Partial Transposition"""

        dims = [dimensions_list, dimensions_list]
        nsys = len(mask)
        pt_dims = np.arange(2 * nsys).reshape(2, nsys).T
        pt_idx = np.concatenate([[pt_dims[n, mask[n]] for n in range(nsys)],
                                [pt_dims[n, 1 - mask[n]] for n in range(nsys)]])

        partial_transpose = rho.reshape(
            np.array(dims).flatten()).transpose(pt_idx).reshape(rho.shape)

        return partial_transpose

def Negativity_Qubits(rho):
        """Compute the witness (PPT) both as value (given the state) and as operator """
        J = (len(rho)-1)/4 
        dim = int(2*J+1)
        dimensions_list = [dim,dim]
        mask = [0,1]
        
        rho_pt = Partial_Transpose_Dense(rho, mask, dimensions_list)
        
        eva, eve = linalg.eig(rho_pt)
        eve = np.transpose(eve)
        
        nega = 0
        witness_matrix = np.zeros((dim**2, dim**2),dtype =np.complex128)
        
        for i in range(len(eva)):
            num = 0
            if eva[i] < 0:
                nega += eva[i]
                witness_matrix += np.outer(eve[i], np.conj(eve[i].T))
                num += 1
        
        witness_matrix = Partial_Transpose_Dense(witness_matrix, mask, dimensions_list)
        return nega, witness_matrix
```
